```python
# Import Python Libraries
import osmnx as ox
import geopandas as gpd
import hvplot.pandas
```

```python
# Look for GRIC with OSM
place = "Gila River Indian Community, Arizona, USA"

gila_boundary = ox.geocode_to_gdf(place)

gila_boundary
```

```python
# Plot GRIC
gila_boundary.plot()
```

```python
gila_boundary.hvplot(
    geo=True,
    tiles="EsriImagery",
    alpha=0.3,
    line_color="black",
    line_width=2,
    title="Gila River Indian Community Boundary",
    width=800,
    height=600
)
```

```python
# Get the boundary polygon
gila_polygon = gila_boundary.geometry.iloc[0]
```

```python
water_tags = {
    "natural": "water",
    "waterway": True
}

gila_water = ox.features_from_polygon(
    gila_polygon,
    tags=water_tags
)

gila_water.head()
```

```python
gila_water.plot()
```
